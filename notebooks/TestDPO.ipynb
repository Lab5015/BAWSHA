## Test single acquisition

```python
tek = DPO3014()
tek.set_average_mode()
tek.AVG = 512
tek.write("ACQuire:STOPAfter SEQUENCE")
tek.write("ACQuire:STATE ON")
while(True):
    try:
        tek.query("BUSY?").strip()==0
        readout = tek.acquire()
        break
    except:
        time.sleep(15)
```

```python
plt.scatter(readout["time"], readout["CH1"], marker='.', color='navy')
```
